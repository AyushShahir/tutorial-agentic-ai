```python
from dotenv import load_dotenv
from langchain_google_genai import ChatGoogleGenerativeAI

load_dotenv()
```

# Call LLM using Google Models

```python
query=input("Ask your Question:")
llm=ChatGoogleGenerativeAI(model="gemini-2.5-flash") #create LLM

response=llm.invoke(query) #call LLM
print(response.text)
```

```python
response=llm.invoke([
    ["system", "You are a helpful assistant that answers in short one line"],
    ["human", query]
])
print(response.text)
```

# Difference in outputs because of change in Temperature

```python
prompt = "Write a single tagline for a quirky tea stall. Only output the tagline, nothing else."

llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=0) #create LLM

response = llm.invoke(prompt)
print(response.text)
```

```python
llm = ChatGoogleGenerativeAI(model="gemini-2.5-flash", temperature=1) #create LLM

response = llm.invoke(prompt)
print(response.text)
```

# Call LLM using Groq Models

```python
from langchain_groq import ChatGroq
```

```python
llm=ChatGroq(model="llama-3.3-70b-versatile" , reasoning_format="parsed")

messages = [
    (
        "system",
        "You are a helpful assistant that gives a short answer to the question",
    ),
    (
        "human",
        "How many moons does Jupiter have?"
    ),
]
ai_msg=llm.invoke(messages)
print(ai_msg.content)
```
